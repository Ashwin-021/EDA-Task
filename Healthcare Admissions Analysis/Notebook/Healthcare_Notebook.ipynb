```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/healthcare_preprocessed (1).csv")
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,
            x="Age",
            bins=12,
            kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("number of patients")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,
            x="Age",
            hue="Gender",
            bins=15,
            kde=True,
             element="step"
             )
plt.title("Patient age distribution by gender")
plt.xlabel("Age")
plt.ylabel("number of patients")
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()
plt.figure(figsize=(8, 6))
sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values

)

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
condition_counts = df['Medical_Condition'].value_counts()
plt.figure(figsize=(8, 6))
sns.barplot(
    x=condition_counts.index,
    y=condition_counts.values

)

plt.title('patient condition Distribution')
plt.xlabel('medical condition')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df,
            x="Medical_Condition",
            hue="Gender"

             )
plt.title("Patient condition distribution by gender")
plt.xlabel("medical condition")
plt.ylabel("number of patients")

plt.xticks(rotation=30)
plt.show()
```

```python
admission_counts = df['Admission_Type'].value_counts()
plt.figure(figsize=(8, 6))
sns.barplot(
    x=admission_counts.index,
    y=admission_counts.values

)

plt.title('Admission Type Distribution')
plt.xlabel('Admission Type')
plt.ylabel('Count')
plt.show()
```

```python
admission_condition=pd.crosstab(df["Medical_Condition"],df["Admission_Type"])
print(admission_condition)
```

```python
admission_condition.plot(kind="bar",figsize=(12,6),stacked=True)
plt.title("Admission Type by Medical Condition")
plt.xlabel("Medical Condition")
plt.legend(title="admission type")
plt.ylabel("Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
 plt.figure(figsize=(10,6))
 sns.histplot(
     data=df,
     x="Billing_Amount",
     bins=15,
     kde=True
 )
 plt.title("Billing Amount Distribution")
 plt.xlabel("Billing Amount")
 plt.ylabel("Count")
 plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(
    data=df,
    x="Medical_Condition",
    y="Billing_Amount"
)
plt.title("Billing Amount Distribution by Medical Condition")
plt.xlabel("Medical Condition")
plt.ylabel("Billing Amount")
plt.xticks(rotation=30)
plt.show()
```

```python
avg_condition_billing = (df.groupby("Medical_Condition")["Billing_Amount"].
                         mean()
                         .sort_values(ascending=False))
print(avg_condition_billing)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x = avg_condition_billing.values,
    y = avg_condition_billing.index,
)
plt.title("Average Billing Amount by Medical Condition")
plt.xlabel("Average Billing Amount")
plt.ylabel("Medical Condition")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="Insurance_Provider",
    y="Billing_Amount"
)

plt.title("Billing Amount Distribution by Insurance Provider")
plt.xlabel("Insurance Provider")
plt.ylabel("Billing Amount")
plt.xticks(rotation=30)
plt.show()
```
